```python
# Pacote Numpy
import numpy as np
```

```python
# Matriz A
A = np.matrix('3 2 1;0 9 2')
A
```

```python
# Transposta da matriz A
A.T
```

```python
# matriz A e matriz B
A = np.matrix('3 0;1 -5') ; B = np.matrix('2 1;3 4')
```

```python
# Soma das matrizes A e B
print(A)
print(B)
print(A+B)
```

```python
# Multiplicação escalar com matriz
print(4*A)
```

```python
# matrizes A e B
A = np.matrix('1 0 3;2 1 4')
B = np.matrix('3 2;1 4;0 1')
print(A)
print(B)
```

```python
# Dimensões das matrizes A e B
print('dim A = ', A.shape, 'e dim B = ', B.shape)
```

```python
# Produto das matrizes A por B
A@B
```

```python
# Produto das matrizes A por B
A.dot(B)
```

```python
# matriz A
A = np.matrix('1 0;2 1')
#Potências de A
print(np.linalg.matrix_power(A,2),'\n')
print(np.linalg.matrix_power(A,3),'\n')
print(np.linalg.matrix_power(A,4),'\n')
print(np.linalg.matrix_power(A,5))
```

```python
# matriz A
A = np.matrix('1 0 1;2 1 1;1 1 1')
print(A)
# inversa da matriz A
print(np.linalg.inv(A))
```

```python
# Determinante da matriz A de ordem 2
A = np.matrix('3 2;1 4')
print(A)
print(int(np.linalg.det(A)))
```

```python
# Determinante da matriz A de ordem 3
A = np.matrix('1 3 -1;3 4 1;1 2 -1')
print(A)
print(int(np.linalg.det(A)))
```
